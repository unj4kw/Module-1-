* Prevalence & incidence
Prevalence refers to the number and proportion of a population living with Alzheimer's dementia at a given time; incidence refers to the rate of new cases diagnosed per year (often expressed per 100,000 people). In 2025, an estimated 7.2 million Americans aged 65+ are living with Alzheimer's dementia, 74% of whom are 75 or older. Prevalence rises sharply with age, from 5.1% among those 65–74 to 33.4% among those 85 and older, and roughly 1 in 9 adults 65+ (11%) is affected overall. Annual incidence follows a similar age gradient: 0.4% (ages 65–74), 3.2% (75–84), and 7.6% (85+). Younger-onset dementia is comparatively rare and understudied, affecting an estimated 110 per 100,000 adults aged 30–64 (~200,000 Americans). Of the 7.2 million affected older adults, 4.4 million (12% of older women) are women and 2.8 million (10% of older men) are men. Prevalence also varies by race, affecting 19% of Black, 14% of Hispanic, and 10% of White older adults. Mild cognitive impairment (MCI), often a precursor to dementia, affects an estimated 22% of adults 65+. Sexual and gender minority (SGM) older adults appear to face elevated dementia risk, with transgender adults 65+ showing a prevalence of 18–21%, compared to 12–13% among cisgender adults — though researchers note that population-level prevalence estimates likely undercount SGM individuals, since studies typically capture only those already engaged with the healthcare system. Given the aging U.S. population — projected to grow from 58 million (65+) in 2022 to 92 million by 2050, with the full baby boom generation reaching 65+ by 2030 — prevalence is expected to rise substantially in the coming decades.

* Economic burden
The total economic burden of Alzheimer's disease and related dementias in the United States is projected to reach $781 billion in 2025. Direct medical and long-term care costs account for $232 billion, while indirect costs include $302 billion in lost quality of life and $233 billion in informal caregiving costs, of which $52 billion is paid out-of-pocket by patients and families. Unpaid family caregivers provided an estimated $413.5 billion in economic value of care in 2024 alone, with unpaid care exceeding $5 billion in each of 27 states. Globally, 68% of the projected increase in dementia prevalence and burden by 2050 is expected to occur in low- and middle-income countries, reflecting demographic shifts and limited healthcare infrastructure in these regions.

* Risk factors (genetic, lifestyle) 
Advancing age remains the strongest known risk factor for Alzheimer's dementia. Sex-based differences in risk appear to be shaped by a combination of biological and sociocultural factors: some multi-country studies report higher incidence among women, particularly in low- and middle-income countries and in Africa and South America, while patterns are less consistent in high-income regions. Lower educational attainment among women born in the early-to-mid 20th century may partly explain this disparity, as limited formal education is itself an independent risk factor for dementia. The APOE-ε4 allele, the best-established common genetic risk factor for Alzheimer's, appears to confer greater risk in women than men, particularly between ages 55–60, suggesting an interaction between genetic and hormonal factors. Modifiable lifestyle risk factors include hypertension, smoking, and midlife obesity.

* Societal determinants
 Lower educational attainment is associated with poorer brain health and elevated dementia risk; rising educational attainment among U.S. women over recent decades has correspondingly reduced dementia risk in this group more than among men. Early workforce participation is similarly protective — women employed earlier in life show better cognitive outcomes after age 60 than women outside the paid workforce. Racial disparities in dementia risk between Hispanic, Black, and White populations are largely attributable to differences in socioeconomic status and comorbid health conditions (e.g., cardiovascular disease, diabetes), compounded by the cumulative physiological effects of social inequality. Additional social determinants include HIV/AIDS (which can independently cause dementia), discrimination and social exclusion faced by SGM older adults, inequitable healthcare access and a shortage of culturally competent providers, environmental exposures (air pollution, unsafe housing, limited green space, poor transportation access), and adverse life experiences such as childhood trauma, sibling loss, late-life financial abuse, and spousal abuse.

* Symptoms
Alzheimer's disease first damages neurons in brain regions responsible for memory, language, and cognition, producing early symptoms such as difficulty recalling recent conversations, names, or events. As the disease progresses, individuals commonly develop communication difficulties, confusion, impaired judgment, and behavioral or personality changes, including apathy, mood shifts, and heightened suspicion or fear. Patients may also lose the ability to plan, follow multi-step tasks, or work with numbers, and may experience disorientation to time and place, along with impaired visuospatial processing that affects tasks like driving. In late stages, motor symptoms such as difficulty walking, speaking, and swallowing become common. Symptoms worsen progressively over time, consistent with the disease's neurodegenerative course.

* Diagnosis
Diagnostic workup begins with a physical and neurological exam assessing reflexes, muscle tone and strength, mobility, sensory function, and coordination. Structural MRI can reveal regional brain atrophy characteristic of AD, while FDG-PET imaging identifies areas of reduced glucose metabolism, a pattern that helps distinguish AD from other dementia subtypes. Tau-PET imaging directly visualizes neurofibrillary tangle burden, and diffusion tensor imaging (DTI) serves as an additional marker of neurodegeneration. Emerging blood-based biomarker tests, such as PrecivityAD, detect specific beta-amyloid species in serum, which are characteristically reduced in AD patients.

* Standard of care treatments (& reimbursement) 
Two FDA-approved disease-modifying therapies, lecanemab (Leqembi®) and donanemab (Kisunla™), target and clear beta-amyloid from the brain, slowing underlying disease progression. Aducanumab (Aduhelm®), approved in 2021 as the first such treatment, was discontinued in 2024 as its manufacturer shifted production priority to Leqembi®. Brexpiprazole is used off-label to manage agitation, likely through modulation of dopaminergic and serotonergic signaling. Non-pharmacological interventions — including physical activity, reminiscence therapy, music and art therapy, pet therapy, and light therapy — are widely used to maintain cognitive and functional health, quality of life, and independence in activities of daily living.  

* Disease progression & prognosis
Following diagnosis, individuals aged 65+ survive an average of four to eight years, though survival can extend beyond 20 years in some cases. Alzheimer's disease is currently estimated to affect 6.2 million Americans and causes more deaths annually than breast and prostate cancer combined. Disease progression is commonly staged using a seven-point continuum: Stages 1–3 represent preclinical AD (no cognitive decline, subjective decline, and mild cognitive impairment, respectively), while Stages 4–7 span mild, moderate, severe, and very severe dementia. Time from diagnosis to advanced stages typically ranges from 3 to 11 years, though some individuals survive 20+ years post-diagnosis.

* Continuum of care providers
Primary care physicians and advanced practice providers typically serve as the initial point of contact and coordinate care across the disease course, working alongside specialists such as neurologists and geriatricians. As functional needs increase, residential care facilities provide support with activities of daily living for individuals who can no longer be safely cared for at home.  

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
Alzheimer's disease is characterized by two hallmark neuropathological changes: extracellular accumulation of beta-amyloid protein fragments into plaques, and intracellular accumulation of hyperphosphorylated tau protein into neurofibrillary tangles. These proteinopathies are accompanied by chronic neuroinflammation and impaired cerebral glucose metabolism. Cumulative neuronal damage and synaptic loss drive progressive brain atrophy — a measurable reduction in brain volume — reflecting the disease's broader disruption of cellular function and systemic physiology over time.

* Clinical Trials/next-gen therapies
Clinical trials of lecanemab demonstrated moderate slowing of cognitive and functional decline in patients with MCI or mild AD-related dementia who had confirmed beta-amyloid accumulation via imaging or CSF analysis. Donanemab trials showed more pronounced slowing of disease progression in a similar population with both amyloid and tau pathology. Neither drug is curative, and both carry eligibility restrictions limiting their use to specific patient populations. As of January 2024, approximately 30 active clinical trials were investigating novel agents targeting the cognitive, behavioral, and neuropsychiatric symptoms of AD. Among emerging candidates, trontinemab — an anti-amyloid monoclonal antibody engineered with "Brainshuttle" technology — is designed to more efficiently cross the blood-brain barrier via transferrin receptor-mediated transport.

```python
# Figure 1: Scatter plots of MMSE vs. each protein, with linear regression.
# A negative slope indicates more of that protein goes with worse cognitive scores.

q1_results = {} # Stores each protein's slope and R^2 so we can compare them and use them in the outlier analysis.

# One figure with a 2x2 grid of scatter plots so all 4 proteins can be compared side by side.
# figsize is large and constrained_layout spaces the panels so titles and labels don't overlap.
fig, axes = plt.subplots(2, 2, figsize = (13, 10), constrained_layout = True)

# Loops through the 4 proteins
# axes.flat turns the 2x2 grid into a list, so each protein gets its own panel.
for ax, (attr, label) in zip(axes.flat, proteins):
    x, y = paired_values(patients, attr, 'mmse') # x = protein level, y = MMSE score
    slope, intercept, r2 = scatter_with_regression(ax, x, y, f'{label} (pg/ug)', 'Last MMSE Score', f'MMSE vs. {label} (n = {len(x)})')
    q1_results[label] = (slope, r2) # Saves the results for this protein.

fig.suptitle('Figure 1: Cognitive Score (MMSE) vs. Protein Level', fontsize = 15) # Main title over the whole figure.
plt.savefig(f'fig1_mmse_vs_proteins.png', dpi = 100) # Saves the figure as a .png file.

# Prints a summary table of the slope and R^2 for each protein.
# The protein with the highest R^2 explains the most variation in MMSE, meaning it tracks cognition best.
# :<9 and :>10 line up the columns, and :.4f / :.3f round to 4 or 3 decimal places.
print(f"{'Protein':<9}{'slope':>10}{'R^2':>8}")
for label, (slope, r2) in q1_results.items():
    print(f"{label:<9}{slope:>10.4f}{r2:>8.3f}")
```

```python
# Figure 2: Bar graphs comparing protein levels in donors with verusus without dementia.
# Null hypothesis: donors with and without dementia have similar levels of the protein.
# Alternative (two-tailed): the levels differ.

# One figure with 2 bar graphs side by side (pTAU and ABeta42).
fig, axes = plt.subplots(1, 2, figsize = (13, 6), constrained_layout = True)

for ax, (attr, label) in zip(axes, [('ptau', 'pTAU'), ('abeta42', 'ABeta42')]):
# Builds two lists of protein values: one for patients with dementia and one for patients without.
# getattr(p, attr) gets the protein value, and patients missing it are skipped.
    dem = [getattr(p, attr) for p in patients if p.cognitive_status == 'Dementia' and getattr(p, attr) is not None]
    no_dem = [getattr(p, attr) for p in patients if p.cognitive_status == 'No dementia' and getattr(p, attr) is not None]
    print(f"{label}:")

    # Draws the bar graph and runs the statistical test on the two groups.
    test_name, stat, p = bar_with_test(ax, [no_dem, dem], ['No dementia', 'Dementia'], f'{label} (pg/ug)', f'{label} by Cognitive Status', ['lightgreen', 'salmon'])
    print(f"{label}: {test_name}, statistic = {stat:.3f}, p = {p:.3g}")

fig.suptitle('Figure 2: Protein Levels in Donors With vs. Without Dementia', fontsize = 15)
plt.savefig(f'fig2_{label}_by_dementia.png', dpi = 100)
```

For pTAU, the data passed the normality and equal-variance checks, so we used a Student’s t-test. pTAU was higher in the dementia group (about 4.7 vs 3.8 pg/ug), but the difference was not statistically significant (p = 0.111).
```python
# Figure 3: One-way ANOVA to visualize if pTAU differs across the 4 levels of AD pathology.
# Null hypothesis: pTAU is the same across all 4 pathology levels.

# The 4 pathology levels, from least to most damage.
levels = ['Not AD', 'Low', 'Intermediate', 'High']

# Builds one list of pTAU values per pathology level using the ad_change filter.
# This creates a list of 4 lists, one for each level.
ptau_groups = [[p.ptau for p in Patient.filter_patients(patients, ad_change = lvl) if p.ptau is not None] for lvl in levels]

# Checks the assumptions first. The * unpacks the 4 lists so each one is passed in as its own group.
if check_assumptions(*ptau_groups):
    # If both checks pass, it runs a one-way ANOVA to compare means.
    anova_name = 'One-way ANOVA'
    stat, anova_p = stats.f_oneway(*ptau_groups)
else:
    # If either check fails, it runs a Kruskal-Wallis ANOVA on ranks.
    anova_name = 'Kruskal-Wallis ANOVA on ranks'
    stat, anova_p = stats.kruskal(*ptau_groups)
print(f"{anova_name}: statistic = {stat:.3f}, p = {anova_p:.3g}\n")

# ANOVA only tells us that at least one group is different, not which one.
# So if it passes (p <= 0.05), Tukey's HSD post hoc test compares every pair of groups to find which ones differ.
if anova_p <= 0.05:
    tukey = stats.tukey_hsd(*ptau_groups)
    print("Tukey HSD post hoc test (pairs with p < 0.05 differ significantly): ")

    # The two loops go through every pair of levels once
    # tukey.pvalue[i, j] is the p-value for the pair of level i and level j
    for i in range(len(levels)):
        for j in range(i + 1, len(levels)):
            print(f" {levels[i]:>12} vs {levels[j]:<12} p = {tukey.pvalue[i,j]:.3g}")
else:
    print("ANOVA failed (p > 0.05), so no post hoc test: none of the groups differ significantly.")

# Bar graph of the mean (+/- stdev) pTAU for each pathology level.
fig, ax = plt.subplots(figsize = (7, 5))
means = [statistics.mean(g) for g in ptau_groups] # bar heights
stds = [statistics.stdev(g) for g in ptau_groups] # error bar sizes

# The 4 colors go from light to dark blue to show increasing pathology.
ax.bar(range(4), means, yerr = stds, capsize = 8, color = ['#d0e1f2', '#9ecae1', '#4292c6', '#08519c'], edgecolor = 'black')
ax.set_xticks(range(4))
ax.set_xticklabels([f'{l}\n(n = {len(g)})' for l, g in zip(levels, ptau_groups)]) # Level name and number of patients under each bar.
ax.set_xlabel('Overall AD Neuropathological Change')
ax.set_ylabel('pTAU (pg/ug)')
ax.set_title(f"pTAU by Alzheimer's Pathology Level\n{anova_name}: p = {anova_p:.3g}") # Shows which test was used and its p-value.
plt.tight_layout() # Adjusts spacing so labels don't get cut off.
plt.savefig('fig3_ptau_anova.png', dpi = 100)
```

Figure 3: We compared pTAU across the four levels of Alzheimer’s pathology. The groups passed the K-S normality check but failed Levene’s test for equal variances because the High group was much more spread out. Therefore, a Kruskal-Wallis ANOVA on ranks was used and was significant (p = 0.004), meaning pTAU differs across pathology levels. Tukey’s HSD post hoc test showed that the only significant pairwise difference was between the Intermediate and High groups (p = 0.007). pTAU remained similar across the Not AD, Low, and Intermediate groups and rose sharply in the High group. This suggests pTAU accumulates mainly at the most advanced stage of Alzheimer’s pathology rather than increasing gradually. The Not AD vs. High (p = 0.106) and Low vs. High (p = 0.073) comparisons were not significant, possibly because of the small sizes of the Not AD and Low groups.

```python
# Outlier Analysis
# Figure 6: The protein data is very skewed: a few donors have extremely high values.
# In linear regression, a single extreme point can pull the line and inflate or shrink R^2.
# Our code finds outliers using the IQR rule, shows them with box plots, and re-runs Figure 1 regressions without the outliers and compare R^2.
# If R^2 stays similar, the result is not driven by a few patients.

print(f"{'Protein':<9}{'# outliers':>11}{'R^2 (all)':>11}{'R^2 (no outliers)':>19}")
for attr, label in proteins:
# All of the values for this protein, skipping missing ones.
    values = [getattr(p, attr) for p in patients if getattr(p, attr) is not None]

    # Gets the outlier cutoffs, then finds every patient whose value falls outside of them.
    low, high_bound = iqr_outlier_bounds(values)
    outliers = [p for p in patients if getattr(p, attr) is not None and not (low <= getattr(p, attr) <= high_bound)]

    # Box plot: the box is the middle 50% of the data, the line inside is the median, and dots past the whiskers are outliers.
    fig, ax = plt.subplots(figsize = (5,5))
    ax.boxplot(values)
    ax.set_title(f'{label} Distribution ({len(outliers)} outliers)')
    ax.set_ylabel(f'{label} (pg/ug)')
    ax.set_xticks([]) # Removes the x-axis tick since there is only one box
    plt.tight_layout()
    plt.savefig(f'fig6_{label}_outliers.png', dpi = 100)

    # Re-runs the regression without the outliers and compares it to the original R^2 from Figure 1.
    kept = [p for p in patients if p not in outliers] # Everyone except the outliers.
    x, y = paired_values(kept, attr, 'mmse')
    _, _, _, r2_clean = linear_regression(x, y) # The _ skips the values we don't need (model, slope, intercept).
    print(f"{label:<9}{len(outliers):>11}{q1_results[label][1]:>11.3f}{r2_clean:>19.3f}")
    print(f"        outlier IDs: {[p.donor_id for p in outliers]}") # Lists which donors were outliers.
```

## Verify and validate your analysis: 

We ran several checks to make sure our results were not caused by data errors, incorrect statistical tests, or extreme values. First, we checked for impossible values. One donor had an MMSE score of 33, even though the maximum score is 30, so we removed this value from all MMSE analyses. Next, before each group comparison, we checked whether the data met the requirements for parametric tests. We used the Kolmogorov-Smirnov test for normality and Levene’s test for equal variances. When the data did not meet these requirements, we used a non-parametric test instead. Due to the protein data being highly skewed, we also checked for outliers using the IQR rule and box plots. We then repeated the Figure 1 regressions without the outliers. This showed that the original ABeta40 relationship with MMSE (R2 = 0.333) was mostly caused by outliers. After removing them, the R2 dropped to 0.086. Without outliers, all four proteins had similarly weak relationships with MMSE (R2 ≈ 0.06-0.09). We also repeated the pTAU and ABeta42 analyses using only donors whose MMSE was measured within 24 months of death (n = 36). pTAU became slightly stronger (R2 = 0.108), while ABeta42 became weaker (R2 = 0.043). Overall, pTAU was the only protein whose negative relationship with MMSE remained consistent across all of these checks.

We compared our results with previous studies. Our Question 1 results generally matched the literature. Previous research found that cognitive decline is more strongly related to tau pathology than amyloid pathology (Nelson et al., 2012). Bennett et al. (2004) also found that amyloid was no longer significantly associated with Alzheimer’s disease after accounting for tau tangles. This supports our finding that pTAU was the most consistent predictor of MMSE. Our ABeta42 result also matched previous research because ABeta42 was significantly higher in donors with dementia (p = 0.002), similar to findings from Bennett et al. (2004). 

However, our Question 2 results did not match the literature. We found no relationship between education and MMSE in the high-pathology group (R2 = 0.000) and no significant difference in education between the dementia and non-dementia groups (p = 0.219). Donors with dementia actually averaged slightly more education (16.4 vs. 15.2 years), the opposite of what cognitive reserve predicts. Bennett et al. (2005), however, found that education affects how strongly brain pathology impacts cognition. This supports the idea of cognitive reserve. The weakness of all our correlations is consistent with Nelson et al. (2012), who note that non-Alzheimer’s brain lesions common in older adults can weaken the link between pathology and cognition.

Several factors may have affected our results. Our high pathology group only had 11 donors without dementia, which made it harder to detect an education effect. Our sample was also highly educated overall, so there was not much variation in education levels. The MMSE is also a short screening test with a maximum score of 30, so it may not capture differences in cognition as well as more detailed cognitive tests. In addition, there was sometimes a large gap between when the MMSE was measured and when the brain proteins were measured after death. Our regressions also reported R2 but not p-values, so we could not formally test whether each regression slope was significantly different from zero. Finally, we did not account for other brain conditions in the dataset, such as LATE, Lewy bodies, and vascular disease, which can also affect cognition.

## Conclusions and Ethical Implications: 

Our results suggest that pTAU was more consistently linked to cognitive decline than amyloid-beta. pTAU was the only protein whose negative relationship with MMSE stayed consistent after removing outliers and using only more recent MMSE scores. It also increased in donors with the highest levels of pathology. However, all of the relationships were still weak (R² < 0.11), so these results alone do not show that pTAU is a strong predictor of cognitive decline. They do suggest that pTAU and tau-related biomarkers could be worth exploring alongside current amyloid-focused treatments. 

For education, we did not find evidence that higher education protected high-pathology donors from dementia. However, this result may have been affected by our small and highly educated sample. A lack of a significant relationship does not mean that education or mental engagement are unimportant, especially since larger studies have found evidence for cognitive reserve. Finally, our study included 84 brain donors from one region, so the results may not apply to other or more diverse populations. Future research on tau-based diagnostics and treatments should include more diverse groups and consider making these treatments accessible regardless of income or insurance.



## Limitations and Future Work: 

Our analysis had several limitations. The sample was small (84 donors), and only 11 high-pathology donors did not have dementia, which made it harder to detect an education effect. The sample was also highly educated, so there was not much variation in education levels. The MMSE is a short screening test, and some scores were taken years before death, while the proteins were measured after death from only one brain region, the middle temporal gyrus. There were also some statistical limitations. Our regressions reported R² without p-values, and we used Tukey’s HSD as our post hoc test even though it assumes equal variances, which our pTAU groups did not have. We did not control for age or other brain conditions, such as LATE, Lewy bodies, and vascular disease. Finally, because this was an observational study, our results show associations, not cause and effect. Future work should use larger and more diverse groups, control for age and other brain conditions, test whether education changes the relationship between pathology and cognition, and use repeated cognitive testing over time to better understand how tau and amyloid relate to cognitive decline.
