* Prevalence & incidence
Prevalence refers to the number and proportion of a population living with Alzheimer's dementia at a given time; incidence refers to the rate of new cases diagnosed per year (often expressed per 100,000 people). In 2025, an estimated 7.2 million Americans aged 65+ are living with Alzheimer's dementia, 74% of whom are 75 or older. Prevalence rises sharply with age, from 5.1% among those 65–74 to 33.4% among those 85 and older, and roughly 1 in 9 adults 65+ (11%) is affected overall. Annual incidence follows a similar age gradient: 0.4% (ages 65–74), 3.2% (75–84), and 7.6% (85+). Younger-onset dementia is comparatively rare and understudied, affecting an estimated 110 per 100,000 adults aged 30–64 (~200,000 Americans). Of the 7.2 million affected older adults, 4.4 million (12% of older women) are women and 2.8 million (10% of older men) are men. Prevalence also varies by race, affecting 19% of Black, 14% of Hispanic, and 10% of White older adults. Mild cognitive impairment (MCI), often a precursor to dementia, affects an estimated 22% of adults 65+. Sexual and gender minority (SGM) older adults appear to face elevated dementia risk, with transgender adults 65+ showing a prevalence of 18–21%, compared to 12–13% among cisgender adults — though researchers note that population-level prevalence estimates likely undercount SGM individuals, since studies typically capture only those already engaged with the healthcare system. Given the aging U.S. population — projected to grow from 58 million (65+) in 2022 to 92 million by 2050, with the full baby boom generation reaching 65+ by 2030 — prevalence is expected to rise substantially in the coming decades.

* Economic burden
The total economic burden of Alzheimer's disease and related dementias in the United States is projected to reach $781 billion in 2025. Direct medical and long-term care costs account for $232 billion, while indirect costs include $302 billion in lost quality of life and $233 billion in informal caregiving costs, of which $52 billion is paid out-of-pocket by patients and families. Unpaid family caregivers provided an estimated $413.5 billion in economic value of care in 2024 alone, with unpaid care exceeding $5 billion in each of 27 states. Globally, 68% of the projected increase in dementia prevalence and burden by 2050 is expected to occur in low- and middle-income countries, reflecting demographic shifts and limited healthcare infrastructure in these regions.

* Risk factors (genetic, lifestyle) 
Advancing age remains the strongest known risk factor for Alzheimer's dementia. Sex-based differences in risk appear to be shaped by a combination of biological and sociocultural factors: some multi-country studies report higher incidence among women, particularly in low- and middle-income countries and in Africa and South America, while patterns are less consistent in high-income regions. Lower educational attainment among women born in the early-to-mid 20th century may partly explain this disparity, as limited formal education is itself an independent risk factor for dementia. The APOE-ε4 allele, the best-established common genetic risk factor for Alzheimer's, appears to confer greater risk in women than men, particularly between ages 55–60, suggesting an interaction between genetic and hormonal factors. Modifiable lifestyle risk factors include hypertension, smoking, and midlife obesity.

* Societal determinants
 Lower educational attainment is associated with poorer brain health and elevated dementia risk; rising educational attainment among U.S. women over recent decades has correspondingly reduced dementia risk in this group more than among men. Early workforce participation is similarly protective — women employed earlier in life show better cognitive outcomes after age 60 than women outside the paid workforce. Racial disparities in dementia risk between Hispanic, Black, and White populations are largely attributable to differences in socioeconomic status and comorbid health conditions (e.g., cardiovascular disease, diabetes), compounded by the cumulative physiological effects of social inequality. Additional social determinants include HIV/AIDS (which can independently cause dementia), discrimination and social exclusion faced by SGM older adults, inequitable healthcare access and a shortage of culturally competent providers, environmental exposures (air pollution, unsafe housing, limited green space, poor transportation access), and adverse life experiences such as childhood trauma, sibling loss, late-life financial abuse, and spousal abuse.

* Symptoms
Alzheimer's disease first damages neurons in brain regions responsible for memory, language, and cognition, producing early symptoms such as difficulty recalling recent conversations, names, or events. As the disease progresses, individuals commonly develop communication difficulties, confusion, impaired judgment, and behavioral or personality changes, including apathy, mood shifts, and heightened suspicion or fear. Patients may also lose the ability to plan, follow multi-step tasks, or work with numbers, and may experience disorientation to time and place, along with impaired visuospatial processing that affects tasks like driving. In late stages, motor symptoms such as difficulty walking, speaking, and swallowing become common. Symptoms worsen progressively over time, consistent with the disease's neurodegenerative course.

* Diagnosis
Diagnostic workup begins with a physical and neurological exam assessing reflexes, muscle tone and strength, mobility, sensory function, and coordination. Structural MRI can reveal regional brain atrophy characteristic of AD, while FDG-PET imaging identifies areas of reduced glucose metabolism, a pattern that helps distinguish AD from other dementia subtypes. Tau-PET imaging directly visualizes neurofibrillary tangle burden, and diffusion tensor imaging (DTI) serves as an additional marker of neurodegeneration. Emerging blood-based biomarker tests, such as PrecivityAD, detect specific beta-amyloid species in serum, which are characteristically reduced in AD patients.

* Standard of care treatments (& reimbursement) 
Two FDA-approved disease-modifying therapies, lecanemab (Leqembi®) and donanemab (Kisunla™), target and clear beta-amyloid from the brain, slowing underlying disease progression. Aducanumab (Aduhelm®), approved in 2021 as the first such treatment, was discontinued in 2024 as its manufacturer shifted production priority to Leqembi®. Brexpiprazole is used off-label to manage agitation, likely through modulation of dopaminergic and serotonergic signaling. Non-pharmacological interventions — including physical activity, reminiscence therapy, music and art therapy, pet therapy, and light therapy — are widely used to maintain cognitive and functional health, quality of life, and independence in activities of daily living.  

* Disease progression & prognosis
Following diagnosis, individuals aged 65+ survive an average of four to eight years, though survival can extend beyond 20 years in some cases. Alzheimer's disease is currently estimated to affect 6.2 million Americans and causes more deaths annually than breast and prostate cancer combined. Disease progression is commonly staged using a seven-point continuum: Stages 1–3 represent preclinical AD (no cognitive decline, subjective decline, and mild cognitive impairment, respectively), while Stages 4–7 span mild, moderate, severe, and very severe dementia. Time from diagnosis to advanced stages typically ranges from 3 to 11 years, though some individuals survive 20+ years post-diagnosis.

* Continuum of care providers
Primary care physicians and advanced practice providers typically serve as the initial point of contact and coordinate care across the disease course, working alongside specialists such as neurologists and geriatricians. As functional needs increase, residential care facilities provide support with activities of daily living for individuals who can no longer be safely cared for at home.  

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
Alzheimer's disease is characterized by two hallmark neuropathological changes: extracellular accumulation of beta-amyloid protein fragments into plaques, and intracellular accumulation of hyperphosphorylated tau protein into neurofibrillary tangles. These proteinopathies are accompanied by chronic neuroinflammation and impaired cerebral glucose metabolism. Cumulative neuronal damage and synaptic loss drive progressive brain atrophy — a measurable reduction in brain volume — reflecting the disease's broader disruption of cellular function and systemic physiology over time.

* Clinical Trials/next-gen therapies
Clinical trials of lecanemab demonstrated moderate slowing of cognitive and functional decline in patients with MCI or mild AD-related dementia who had confirmed beta-amyloid accumulation via imaging or CSF analysis. Donanemab trials showed more pronounced slowing of disease progression in a similar population with both amyloid and tau pathology. Neither drug is curative, and both carry eligibility restrictions limiting their use to specific patient populations. As of January 2024, approximately 30 active clinical trials were investigating novel agents targeting the cognitive, behavioral, and neuropsychiatric symptoms of AD. Among emerging candidates, trontinemab — an anti-amyloid monoclonal antibody engineered with "Brainshuttle" technology — is designed to more efficiently cross the blood-brain barrier via transferrin receptor-mediated transport.
